```python
import pandas as pd 
import os
import numpy as np


Sepal = pd.read_csv(os.path.expanduser(r"~\Downloads\Sepal_Data.csv"))

Petal = pd.read_csv(os.path.expanduser(r"~\Downloads\Petal_Data.csv"))

Combined = pd.concat([Petal, Sepal], axis=1)

Combined = Combined.loc[:, ~Combined.columns.duplicated()]

Exclude = Combined.drop(columns=["sample_id", "species", "Unnamed: 0"])

Exclude.corr()

Mean = Mean = Exclude.mean()
print(f" The mean is:\n{Mean}")

Median = Exclude.median()
print(f" The median is:\n{Median}")

Standard_Deviation = Exclude.std()
print(f" The standard deviation is:\n{Standard_Deviation}")

Grouped = Combined.groupby("species")[["sepal_length", "sepal_width", "petal_length", "petal_width"]]

Means_by_species = Grouped.mean()
Stds_by_species = Grouped.std()

print(Means_by_species)
print(Stds_by_species)

#The two species with the most similar means are versicolor and virginica, as their mean values for sepal length, sepal width, petal length, and petal width are closer to each other compared to setosa. The two species with the most different means are setosa and virginica, as their mean values for these measurements are more distinct from each other.
```
